```python


input_file = r"C:\Users\Dell\Desktop\Analytics Platform\AI_Powered_Ecommerce_Analytics_Platform_2000_Data_2020_2026_Modified.xlsx"
output_file = r"C:\Users\Dell\Desktop\Analytics Platform\ecommerce_cleaned.csv"

df = pd.read_excel(input_file, sheet_name="Ecommerce_Data")

print("Original shape:", df.shape)

# Remove duplicate rows
df = df.drop_duplicates()

# Convert date
df["Order Date"] = pd.to_datetime(df["Order Date"])

# Convert numerical columns
numeric_columns = [
    "Quantity",
    "Price (INR)",
    "Discount (%)",
    "Rating",
    "Sales/Revenue (INR)"
]

for column in numeric_columns:
    df[column] = pd.to_numeric(df[column], errors="coerce")

# Remove rows with missing values created by conversion
df = df.dropna()

# Reset index
df = df.reset_index(drop=True)

df.to_csv(output_file, index=False)

print("Cleaned shape:", df.shape)
print("Cleaned dataset saved to:", output_file)
```

```python
input_file = r"C:\Users\Dell\Desktop\Analytics Platform\ecommerce_cleaned.csv"
output_file = r"C:\Users\Dell\Desktop\Analytics Platform\ecommerce_features.csv"

df = pd.read_csv(input_file)

df["Order Date"] = pd.to_datetime(df["Order Date"])

# Time-based features
df["Year"] = df["Order Date"].dt.year
df["Month"] = df["Order Date"].dt.month
df["Month Name"] = df["Order Date"].dt.month_name()
df["Quarter"] = df["Order Date"].dt.quarter
df["Day of Week"] = df["Order Date"].dt.day_name()

# Discount amount
df["Discount Amount (INR)"] = (
    df["Quantity"] *
    df["Price (INR)"] *
    df["Discount (%)"] / 100
)

# Gross sales
df["Gross Sales (INR)"] = (
    df["Quantity"] * df["Price (INR)"]
)

# Verify revenue
df["Calculated Revenue (INR)"] = (
    df["Gross Sales (INR)"] -
    df["Discount Amount (INR)"]
)

df.to_csv(output_file, index=False)

print("Feature engineering completed.")
print("Shape:", df.shape)
print("\nNew columns:")
print([
    "Year",
    "Month",
    "Month Name",
    "Quarter",
    "Day of Week",
    "Discount Amount (INR)",
    "Gross Sales (INR)",
    "Calculated Revenue (INR)"
])
```

```python
## CUSTOMER ANALYTICS

import pandas as pd

# Load dataset
df = pd.read_csv(r"C:\Users\Dell\Desktop\Analytics Platform\ecommerce_features.csv")

# Create customer-level analytics
customer_summary = df.groupby("Customer ID").agg({
    "Order ID": "nunique",                # Number of Orders
    "Quantity": "sum",                    # Total Quantity Purchased
    "Sales/Revenue (INR)": "sum",         # Total Spending
    "Rating": "mean"                      # Average Rating
}).reset_index()

# Rename columns
customer_summary.columns = [
    "Customer ID",
    "Number of Orders",
    "Total Quantity",
    "Total Spending",
    "Average Rating"
]

# Calculate Average Order Value
customer_summary["Average Order Value"] = (
    customer_summary["Total Spending"] /
    customer_summary["Number of Orders"]
)

# Reorder columns
customer_summary = customer_summary[
    [
        "Customer ID",
        "Number of Orders",
        "Total Quantity",
        "Total Spending",
        "Average Order Value",
        "Average Rating"
    ]
]

# Sort by Total Spending
customer_summary = customer_summary.sort_values(
    by="Total Spending",
    ascending=False
)

# Save output
customer_summary.to_csv(
    r"C:\Users\Dell\Desktop\Analytics Platform\customer_analytics.csv",
    index=False
)

print("=" * 60)
print("CUSTOMER ANALYTICS GENERATED")
print("=" * 60)

print(customer_summary.head(10))

print("\nTotal Customers:", customer_summary.shape[0])

print("\nFile Saved:")
print(r"C:\Users\Dell\Desktop\Analytics Platform\customer_analytics.csv")

```

```python
## SALES TREND ANALYSIS


import matplotlib.pyplot as plt
import os

# Load dataset
df = pd.read_csv(r"C:\Users\Dell\Desktop\Analytics Platform\ecommerce_features.csv")

# Convert Order Date to datetime
df["Order Date"] = pd.to_datetime(df["Order Date"])

# Create output folders
os.makedirs("outputs/charts", exist_ok=True)
os.makedirs("outputs/results", exist_ok=True)

# ==========================================================
# 1. DAILY SALES
# ==========================================================

daily_sales = (
    df.groupby("Order Date")["Sales/Revenue (INR)"]
    .sum()
    .reset_index()
)

daily_sales.columns = ["Order Date", "Total Sales"]

daily_sales.to_csv(r"C:\Users\Dell\Desktop\Analytics Platform\daily_sales.csv",
    index=False
)

print("\nDAILY SALES")
print(daily_sales.head(10))


# ==========================================================
# 2. MONTHLY SALES
# ==========================================================

monthly_sales = (
    df.groupby(["Year", "Month"])["Sales/Revenue (INR)"]
    .sum()
    .reset_index()
)

monthly_sales["Month-Year"] = (
    monthly_sales["Year"].astype(str)
    + "-"
    + monthly_sales["Month"].astype(str).str.zfill(2)
)

monthly_sales.to_csv(
    r"C:\Users\Dell\Desktop\Analytics Platform\monthly_sales.csv",
    index=False
)

print("\nMONTHLY SALES")
print(monthly_sales)


# ==========================================================
# 3. CATEGORY-WISE SALES
# ==========================================================

category_sales = (
    df.groupby("Product Category")["Sales/Revenue (INR)"]
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)

category_sales.columns = [
    "Product Category",
    "Total Sales"
]

category_sales.to_csv(
    r"C:\Users\Dell\Desktop\Analytics Platform\category_sales.csv",
    index=False
)

print("\nCATEGORY-WISE SALES")
print(category_sales)


# ==========================================================
# 4. LOCATION-WISE SALES
# ==========================================================

location_sales = (
    df.groupby("Customer Location")["Sales/Revenue (INR)"]
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)

location_sales.columns = [
    "Customer Location",
    "Total Sales"
]

location_sales.to_csv(
    r"C:\Users\Dell\Desktop\Analytics Platform\location_sales.csv",
    index=False
)

print("\nLOCATION-WISE SALES")
print(location_sales)


# ==========================================================
# 5. MONTHLY SALES TREND CHART
# ==========================================================

plt.figure(figsize=(10, 5))

plt.plot(
    monthly_sales["Month-Year"],
    monthly_sales["Sales/Revenue (INR)"],
    marker="o"
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales (INR)")
plt.xticks(rotation=45)
plt.tight_layout()

plt.savefig(
    r"C:\Users\Dell\Desktop\Analytics Platform\monthly_sales_trend.png"
)

plt.close()


# ==========================================================
# 6. CATEGORY SALES CHART
# ==========================================================

plt.figure(figsize=(10, 5))

plt.bar(
    category_sales["Product Category"],
    category_sales["Total Sales"]
)

plt.title("Sales by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Sales (INR)")
plt.xticks(rotation=45)
plt.tight_layout()

plt.savefig(
    r"C:\Users\Dell\Desktop\Analytics Platform\category_sales.png"
)

plt.close()


# ==========================================================
# 7. LOCATION SALES CHART
# ==========================================================

plt.figure(figsize=(10, 5))

plt.bar(
    location_sales["Customer Location"],
    location_sales["Total Sales"]
)

plt.title("Sales by Customer Location")
plt.xlabel("Customer Location")
plt.ylabel("Sales (INR)")
plt.xticks(rotation=45)
plt.tight_layout()

plt.savefig(
   r"C:\Users\Dell\Desktop\Analytics Platform\location_sales.png"
)

plt.close()


print("\n===================================")
print("SALES ANALYSIS COMPLETED")
print("===================================")

print("\nFiles created:")
print("1. outputs/results/daily_sales.csv")
print("2. outputs/results/monthly_sales.csv")
print("3. outputs/results/category_sales.csv")
print("4. outputs/results/location_sales.csv")

print("\nCharts created:")
print("1. outputs/charts/monthly_sales_trend.png")
print("2. outputs/charts/category_sales.png")
print("3. outputs/charts/location_sales.png")
```
